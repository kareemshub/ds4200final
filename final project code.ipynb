```python
import pandas as pd
import altair as alt

# Load your dataset (replace 'your_data.csv' with actual file)
aid = pd.read_csv('us_foreign_aid_complete.csv')

# Clean 'Fiscal Year' column
aid['Fiscal Year'] = pd.to_numeric(aid['Fiscal Year'], errors='coerce')
aid = aid.dropna(subset=['Fiscal Year'])
aid['Fiscal Year'] = aid['Fiscal Year'].astype(int)

selected_columns = [
    'Country ID', 'Country Code', 'Country Name', 'Region ID', 'Region Name', 
    'Income Group ID', 'Income Group Name', 'International Category ID', 'International Category Name',
    'International Sector Code', 'International Sector Name', 'International Purpose Code', 'International Purpose Name',
    'US Category ID', 'US Category Name', 'US Sector ID', 'US Sector Name', 'Foreign Assistance Objective ID',
    'Foreign Assistance Objective Name', 'Aid Type Group ID', 'Aid Type Group Name', 'Activity ID', 'Submission ID',
    'Activity Name', 'Activity Description', 'Activity Project Number', 'Activity Start Date', 'Activity End Date',
    'Transaction Type ID', 'Transaction Type Name', 'Fiscal Year', 'Transaction Date', 'Current Dollar Amount', 
    'Constant Dollar Amount', 'aid_type_id', 'aid_type_name', 'activity_budget_amount', 'submission_activity_id'
]

aid2 = aid[selected_columns]

# Aggregate total funding per international purpose
aid_agg = aid2.groupby('International Purpose Name', as_index=False)['Current Dollar Amount'].sum()

# Get the top 10 international purposes by total funding
top_10_purposes = aid_agg.nlargest(10, 'Current Dollar Amount')['International Purpose Name']

# Filter the original dataset for only the top 10 purposes
aid_filter = aid[aid['International Purpose Name'].isin(top_10_purposes)]

# Aggregate again per year for visualization
aid_plot = aid_filter.groupby(['Fiscal Year', 'International Purpose Name'], as_index=False)['Current Dollar Amount'].sum()

# Create the stacked bar chart
chart = alt.Chart(aid_plot).mark_bar().encode(
    x=alt.X('Fiscal Year:O', title='Year'),
    y=alt.Y('sum(Current Dollar Amount):Q', title='Total Dollar Amount'),
    color=alt.Color('International Purpose Name:N', title='Top 10 International Purposes'),
    tooltip=['Fiscal Year', 'International Purpose Name', 'sum(Current Dollar Amount)']
).properties(
    title="Top 10 Foreign Aid Purposes by Funding per Year",
    width=800,
    height=500
).interactive()

chart
```

```python
# Aggregate total funding per year
aid_plot = aid2.groupby(['Fiscal Year'], as_index=False).agg(
    total_funding=('Current Dollar Amount', 'sum'),
    top_purpose=('International Purpose Name', lambda x: x.value_counts().idxmax())
)

year_range = alt.selection_interval(encodings=['x'])  # Select range on x-axis

# Create the bar chart
chart = alt.Chart(aid_plot).mark_bar().encode(
    x=alt.X('Fiscal Year:O', title='Year'),
    y=alt.Y('total_funding:Q', title='Total Dollar Amount'),
    tooltip=['Fiscal Year', 'top_purpose', 'total_funding']
).add_params(
    year_range
).properties(
    title="Total Foreign Aid Funding per Year with Top Purpose",
    width=800,
    height=500
).interactive()

chart
```

```python
import altair as alt
import pandas as pd
import numpy as np

# Load and prepare the data
df = pd.read_csv('us_foreign_aid_country.csv')

# Filter for European and Asian countries
regions = ['Europe and Eurasia', 'South and Central Asia', 'East Asia and Oceania']
europe_asia_df = df[df['Region Name'].isin(regions)]

# Clean and convert Fiscal Year to integer
def clean_year(year_str):
    try:
        return int(str(year_str)[:4])
    except:
        return np.nan

europe_asia_df['Fiscal Year'] = europe_asia_df['Fiscal Year'].apply(clean_year)
europe_asia_df = europe_asia_df.dropna(subset=['Fiscal Year'])

# Aggregate by region and year
aid_by_region = europe_asia_df.groupby(['Region Name', 'Fiscal Year'])['current_amount'].sum().reset_index()

# Custom color scheme
region_colors = {
    'Europe and Eurasia': '#1f77b4',
    'South and Central Asia': '#ff7f0e',
    'East Asia and Oceania': '#2ca02c'
}

# Add 'All' option for selection menu
region_options = list(region_colors.keys()) + ['All']

# Set up selection
selection = alt.selection_point(
    name='RegionSelector',
    fields=['Region Name'],
    bind=alt.binding_radio(options=region_options, name='Region: '),
    value='All'
)

# Get year ticks
min_year = int(aid_by_region['Fiscal Year'].min())
max_year = int(aid_by_region['Fiscal Year'].max())
year_ticks = list(range(min_year, max_year + 1, 10))

# Build chart with working filter logic
scatter_plot = alt.Chart(aid_by_region).transform_filter(
    # Expression: if 'All' selected, return true for all; otherwise match
    "RegionSelector['Region Name'] == 'All' || RegionSelector['Region Name'] == datum['Region Name']"
).mark_circle(size=60).encode(
    x=alt.X('Fiscal Year:O', title='Year', axis=alt.Axis(values=year_ticks)),
    y=alt.Y('current_amount:Q', title='Aid Amount (USD)', scale=alt.Scale(type='log')),
    color=alt.Color('Region Name:N',
                    scale=alt.Scale(domain=list(region_colors.keys()),
                                    range=list(region_colors.values())),
                    legend=alt.Legend(title='Region', labelFontSize=12, titleFontSize=14)),
    opacity=alt.value(0.9),
    tooltip=['Region Name', 'Fiscal Year', 'current_amount']
).properties(
    width=800,
    height=400,
    title='US Foreign Aid Over Time by Region (Select Region Below)',
    background='beige'  # Background for the chart container
).configure_view(
    stroke=None,
    fill='beige'  # Match inner chart background
).configure_axis(
    gridColor="lightgray",
    labelAngle=45
).configure_axisX(grid=False).configure_axisY(grid=False).add_params(selection)

scatter_plot.save('USAID Over Time by Region (interactive).html')

scatter_plot
```

```python
import pandas as pd
import altair as alt

# Load and clean data
df = pd.read_csv('us_foreign_aid_funding.csv')
df = df[['Country Name', 'Funding Account Name', 'Transaction Type Name', 'Fiscal Year', 'current_amount']]
df = df[df['current_amount'] > 0]

# Assign regions
def assign_region(country):
    europe_eurasia = ['Albania', 'Armenia', 'Azerbaijan', 'Belarus', 'Bosnia and Herzegovina',
                      'Bulgaria', 'Croatia', 'Czechia', 'Estonia', 'Georgia', 'Hungary',
                      'Kazakhstan', 'Kosovo', 'Moldova', 'Montenegro', 'North Macedonia',
                      'Russia', 'Serbia', 'Ukraine']
    east_asia_oceania = ['China', 'Indonesia', 'Cambodia', 'Timor-Leste']
    south_central_asia = ['Afghanistan', 'Bangladesh', 'India', 'Pakistan']
    if country in europe_eurasia:
        return 'Europe & Eurasia'
    elif country in east_asia_oceania:
        return 'East Asia & Oceania'
    elif country in south_central_asia:
        return 'South & Central Asia'
    return None

df['Region'] = df['Country Name'].apply(assign_region)
df = df[df['Region'].notna()]

# Assign sectors

# Categorize funding by sector
def assign_sector(account_name):
    account_name = str(account_name).lower()

    if 'capital' in account_name or 'investment' in account_name:
        return 'Infrastructure'
    elif 'stabilization' in account_name or 'security' in account_name:
        return 'Security'
    elif 'health' in account_name or 'medical' in account_name or 'hiv' in account_name:
        return 'Health'
    elif 'education' in account_name or 'school' in account_name:
        return 'Education'
    elif 'democracy' in account_name or 'governance' in account_name:
        return 'Democracy & Governance'
    
    # Categorizing uncategorized items into meaningful labeled groups
    humanitarian = [
        'assistance for europe, eurasia and central asia (aeeca)', 
        'department of state, migration and refugee assistance',
        'u.s. emergency refugee and migration assistance fund',
        'tsunami recovery and reconstruction fund'
    ]
    
    economic_dev = [
        'european bank for reconstruction and development',
        'contribution to the international bank for reconstruction and development',
        'agency for international development, development assistance',
        'economic support fund and development fund',
        'millennium challenge corporation, expenses'
    ]
    
    environmental = [
        'forest and rangeland research, forest service',
        'state and private forestry, forest service',
        'national forest system, forest service',
        'u.s. fish and wildlife service, resource management',
        'u.s. fish and wildlife service, multinational species conservation fund',
        'u.s. fish and wildlife service, contributed funds',
        'environmental protection agency, science and technology',
        'environmental protection agency, environmental programs and management'
    ]
    
    agricultural = [
        'commodity credit corporation, export loans program account',
        'agricultural research service, salaries and expenses',
        'national institute of food and agriculture, integrated activities',
        'public law 480 title i food for progress credits, program account',
        'public law 480 title i direct credit and food for progress program account',
        'public law 480 title ii grants',
        'foreign agricultural service, salaries and expenses',
        'commodity credit corporation fund',
        'agricultural research service, miscellaneous contributed funds'
    ]
    
    law_enforcement = [
        'military assistance programs (old code)',
        'international narcotics control and law enforcement',
        'peace keeping operations',
        'foreign military financing program',
        'drug interdiction and counter-drug activities',
        'department of defense, cooperative threat reduction account',
        'presidential drawdown authority',
        'excess defense articles',
        'assets forfeiture fund'
    ]
    
    health_services = [
        'centers for disease control, disease control, research, and training',
        'centers for disease control and prevention, emerging and zoonotic infectious diseases',
        'centers for disease control and prevention, immunization and respiratory diseases',
        'food and drug administration',
        'department of state, global hiv/aids initiative'
    ]
    
    defense = [
        'department of defense, operation and maintenance, defense-wide',
        'department of defense, overseas contingency operations transfer fund',
        'department of defense, overseas humanitarian, disaster, and civic aid',
        'department of defense - unknown funding account',
        'department of the army, operations and maintenance',
        'department of the army, military construction',
        'department of the army, pakistan counterinsurgency fund',
        'department of the army, afghanistan infrastructure fund',
        'department of the navy, operation and maintenance',
        'department of the air force, operations and maintenance',
        'department of the air force, air force cadet fund'
    ]
    
    trade_and_diplomacy = [
        'department of state, diplomatic and consular programs',
        'department of state, international organizations and programs',
        'agency for international development, international assistance program, transition initiatives',
        'trade and development agency',
        'peace corps'
    ]
    
    administration = [
        'agency for international development, operating expenses',
        'agency for international development, office of inspector general, operating expenses',
        'agency for international development, complex crises fund',
        'agency for international development, development credit authority',
        'sovereign credit program account',
        'agency for international development, property management fund',
        'foreign national employees separation liability fund',
        'agency for international development, gifts and donations',
        'open world leadership center trust fund'
    ]
    
    infrastructure = [
        'federal aviation administration, operations',
        'grants-in-aid for airports - recovery act, federal aviation administration, transportation',
        'national institute of standards and technology, scientific and technical research and services',
        'patent and trademark office, salaries and expenses',
        'international trade administration, operations and administration',
        'national oceanic and atmospheric administration, operations, research and facilities'
    ]
    
    economic_support = [
        'department of energy, defense nuclear nonproliferation',
        'department of energy, energy efficiency and renewable energy, energy programs',
        'federal trade commission, salaries and expenses',
        'national science foundation, research and related activities'
    ]
    
    justice = [
        'department of justice, general legal activities',
        'federal bureau of investigation, salaries and expenses',
        'interagency drug enforcement, drug enforcement administration',
        'united states marshals service, salaries and expenses',
        'bureau of alcohol, tobacco, firearms and explosives, salaries and expenses',
        'drug enforcement administration, salaries and expenses'
    ]

    # Check if account_name matches any category, else default to Uncategorized
    account_name_lower = account_name.lower()
    
    if any(group.lower() in account_name_lower for group in humanitarian):
        return 'Humanitarian Assistance'
    elif any(group.lower() in account_name_lower for group in economic_dev):
        return 'Economic Development & Reconstruction'
    elif any(group.lower() in account_name_lower for group in environmental):
        return 'Environmental Protection'
    elif any(group.lower() in account_name_lower for group in agricultural):
        return 'Agriculture & Rural Development'
    elif any(group.lower() in account_name_lower for group in law_enforcement):
        return 'Law Enforcement & Defense'
    elif any(group.lower() in account_name_lower for group in health_services):
        return 'Health & Medical Services'
    elif any(group.lower() in account_name_lower for group in defense):
        return 'Defense & Military'
    elif any(group.lower() in account_name_lower for group in trade_and_diplomacy):
        return 'Trade & Diplomatic Affairs'
    elif any(group.lower() in account_name_lower for group in administration):
        return 'Administrative & Operational Funds'
    elif any(group.lower() in account_name_lower for group in infrastructure):
        return 'Infrastructure & Transportation'
    elif any(group.lower() in account_name_lower for group in economic_support):
        return 'Economic Support & Research'
    elif any(group.lower() in account_name_lower for group in justice):
        return 'Justice & Legal Affairs'
    
    return 'Uncategorized'  # Default to Uncategorized if no match


df['Sector'] = df['Funding Account Name'].apply(assign_sector)
df = df[df['Sector'] != 'Uncategorized']

# Aggregate
agg_df = df.groupby(['Fiscal Year', 'Region', 'Sector'])['current_amount'].sum().reset_index()

# Filter to every 5 years (e.g., 2000, 2005, ...)
years = sorted(agg_df['Fiscal Year'].unique())
filtered_years = [y for y in years if y % 5 == 0]
agg_df = agg_df[agg_df['Fiscal Year'].isin(filtered_years)]

# Plot
chart = alt.Chart(agg_df).mark_bar().encode(
    x=alt.X('Fiscal Year:O', title='Fiscal Year'),
    y=alt.Y('current_amount:Q', title='Aid Amount (USD)'),
    color=alt.Color('Sector:N', title='Sector'),
    column=alt.Column('Region:N', title='Region'),
    tooltip=['Region', 'Sector', 'Fiscal Year', 'current_amount']
).properties(
    width=200,
    height=400,
    title='U.S. Foreign Aid by Sector and Region (Every 5 Years)',
    background='beige'  # Background for the chart container
)

chart.configure_axis(labelAngle=0)

chart.save('USAID by Sector and Region (interactive).html')

chart
```

```python
import pandas as pd
import altair as alt
import numpy as np

# 
foreignaid = pd.read_csv('us_foreign_aid_country.csv')
foreignsectoraid = pd.read_csv('us_foreign_aid_usg_sector.csv')
foreignaid.head
```

```python
mena_subsaharan = foreignaid[foreignaid['Region Name'].isin(['Middle East and North Africa', 'Sub-Saharan Africa'])].copy()

def clean_year(year_str):
    try:
        return int(str(year_str)[:4])
    except:
        return np.nan

mena_subsaharan['Fiscal Year'] = mena_subsaharan['Fiscal Year'].apply(clean_year)
mena_subsaharan = mena_subsaharan.dropna(subset=['Fiscal Year', 'current_amount'])
mena_subsaharan = mena_subsaharan[mena_subsaharan['current_amount'] > 0]

# Fiscal years: 2015–2023
mena_subsaharan = mena_subsaharan[
    (mena_subsaharan['Fiscal Year'] >= 2015) &
    (mena_subsaharan['Fiscal Year'] <= 2023)
]

# Remove outliers
q1 = mena_subsaharan['current_amount'].quantile(0.01)
q99 = mena_subsaharan['current_amount'].quantile(0.99)
mena_subsaharan = mena_subsaharan[
    (mena_subsaharan['current_amount'] > q1) & 
    (mena_subsaharan['current_amount'] < q99)
]

# Build the base chart
base = alt.Chart(mena_subsaharan).mark_boxplot(extent='min-max').encode(
    x=alt.X('Fiscal Year:O', title='Fiscal Year'),
    y=alt.Y('current_amount:Q', title='Aid Amount (USD)', scale=alt.Scale(type='log')),
    color=alt.Color('Region Name:N', title='Region'),
    tooltip=['Region Name', 'Fiscal Year', 'current_amount']
).properties(
    width=350,
    height=400
)

# Facet the chart by region
faceted = base.facet(
    column=alt.Column('Region Name:N', title=None)
).properties(
    title='US Foreign Aid to MENA & Sub-Saharan Africa (2015–2023)',
    background='beige'
)

final_chart = faceted.configure_view(
    strokeWidth=0
).configure_axis(
    grid=False,
    labelAngle=45
)

final_chart.save('USAID to MENA & SSA (interactive).html')
```
